### 1. IMPORT LIBRARIES


```python
# ============================================================
# 1. IMPORT LIBRARIES
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os

# Display settings
pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)

# Plot settings
sns.set_theme(style="whitegrid")

print("Libraries imported successfully.")
```

### 2. LOAD DATASET


```python
# ============================================================
# 2. LOAD DATASET
# ============================================================

# Auto-detect path (Google Colab / Local)
possible_paths = [
    "/content/hospital_readmission_dataset.csv",
    "hospital_readmission_dataset.csv",
    "../data/raw/hospital_readmission_dataset.csv",
    "data/raw/hospital_readmission_dataset.csv"
]

data_path = next((p for p in possible_paths if os.path.exists(p)), "/content/hospital_readmission_dataset.csv")

df = pd.read_csv(data_path)

print(f"Dataset loaded successfully from: {data_path}")
print(f"Total records: {len(df):,}")
```

### 3. BASIC DATASET VERIFICATION


```python
# ============================================================
# 3. BASIC DATASET VERIFICATION
# ============================================================

print("=" * 60)
print("DATASET OVERVIEW")
print("=" * 60)

print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

print("\nFirst 5 records:")
display(df.head())

print("\nColumn names:")
print(df.columns.tolist())

print("\nData types:")
display(df.dtypes)

print("\nMissing values:")
display(df.isnull().sum())

print("\nTotal missing values:", df.isnull().sum().sum())

print("\nDuplicate records:", df.duplicated().sum())
```

### 4. NUMERICAL SUMMARY


```python
# ============================================================
# 4. NUMERICAL SUMMARY
# ============================================================

print("=" * 60)
print("NUMERICAL FEATURES SUMMARY")
print("=" * 60)

display(df.describe())
```

### 5. CATEGORICAL SUMMARY


```python
# ============================================================
# 5. CATEGORICAL SUMMARY
# ============================================================

print("=" * 60)
print("CATEGORICAL FEATURES SUMMARY")
print("=" * 60)

display(df.describe(include="object"))
```

### 6. TARGET VARIABLE ANALYSIS


```python
# ============================================================
# 6. TARGET VARIABLE ANALYSIS
# ============================================================

print("=" * 60)
print("TARGET VARIABLE: LABEL")
print("=" * 60)

print("Target counts:")
display(df["label"].value_counts())

print("\nTarget percentages:")
display(
    (df["label"].value_counts(normalize=True) * 100)
    .round(2)
)

# Target visualization
plt.figure(figsize=(6, 4))

sns.countplot(
    data=df,
    x="label"
)

plt.title("Hospital Readmission Distribution")
plt.xlabel("Readmission (0 = No, 1 = Yes)")
plt.ylabel("Number of Patients")

plt.show()
```

### 7. IDENTIFY FEATURE TYPES


```python
# ============================================================
# 7. IDENTIFY FEATURE TYPES
# ============================================================

numerical_features = df.select_dtypes(
    include=["int64", "float64"]
).columns.tolist()

categorical_features = df.select_dtypes(
    include=["object"]
).columns.tolist()

# Remove target from numerical features
if "label" in numerical_features:
    numerical_features.remove("label")

print("=" * 60)
print("NUMERICAL FEATURES")
print("=" * 60)

print(numerical_features)

print("\n" + "=" * 60)
print("CATEGORICAL FEATURES")
print("=" * 60)

print(categorical_features)
```

### 8. UNIVARIATE ANALYSIS — NUMERICAL FEATURES


```python
# ============================================================
# 8. UNIVARIATE ANALYSIS — NUMERICAL FEATURES
# ============================================================

print("=" * 60)
print("UNIVARIATE ANALYSIS — NUMERICAL FEATURES")
print("=" * 60)

# Statistical summary
display(df[numerical_features].describe())

# Histograms
df[numerical_features].hist(
    figsize=(14, 10),
    bins=15
)

plt.suptitle(
    "Distribution of Numerical Features",
    fontsize=16
)

plt.tight_layout()
plt.show()
```

### 9. UNIVARIATE ANALYSIS — CATEGORICAL FEATURES


```python
# ============================================================
# 9. UNIVARIATE ANALYSIS — CATEGORICAL FEATURES
# ============================================================

print("=" * 60)
print("UNIVARIATE ANALYSIS — CATEGORICAL FEATURES")
print("=" * 60)

for feature in categorical_features:

    print("\n" + "-" * 50)
    print(feature.upper())
    print("-" * 50)

    display(df[feature].value_counts())

# Categorical visualizations
for feature in categorical_features:

    plt.figure(figsize=(8, 4))

    sns.countplot(
        data=df,
        x=feature,
        order=df[feature].value_counts().index
    )

    plt.title(
        f"{feature.replace('_', ' ').title()} Distribution"
    )

    plt.xlabel(
        feature.replace('_', ' ').title()
    )

    plt.ylabel("Number of Patients")

    plt.xticks(rotation=45)

    plt.tight_layout()
    plt.show()
```

### 10. BIVARIATE ANALYSIS — NUMERICAL FEATURES vs TARGET


```python
# ============================================================
# 10. BIVARIATE ANALYSIS — NUMERICAL FEATURES vs TARGET
# ============================================================

print("=" * 60)
print("BIVARIATE ANALYSIS — NUMERICAL FEATURES vs READMISSION")
print("=" * 60)

for feature in numerical_features:

    plt.figure(figsize=(7, 4))

    sns.boxplot(
        data=df,
        x="label",
        y=feature
    )

    plt.title(
        f"{feature.replace('_', ' ').title()} vs Readmission"
    )

    plt.xlabel("Readmission (0 = No, 1 = Yes)")
    plt.ylabel(feature.replace('_', ' ').title())

    plt.tight_layout()
    plt.show()
```

### 11. BIVARIATE ANALYSIS — CATEGORICAL FEATURES vs TARGET


```python
# ============================================================
# 11. BIVARIATE ANALYSIS — CATEGORICAL FEATURES vs TARGET
# ============================================================

print("=" * 60)
print("BIVARIATE ANALYSIS — CATEGORICAL FEATURES vs READMISSION")
print("=" * 60)

for feature in categorical_features:

    print("\n" + "-" * 50)
    print(f"{feature.upper()} vs READMISSION")
    print("-" * 50)

    # Crosstab with percentages
    table = pd.crosstab(
        df[feature],
        df["label"],
        normalize="index"
    ) * 100

    display(table.round(2))

    # Visualization
    plt.figure(figsize=(8, 4))

    sns.countplot(
        data=df,
        x=feature,
        hue="label",
        order=df[feature].value_counts().index
    )

    plt.title(
        f"{feature.replace('_', ' ').title()} vs Readmission"
    )

    plt.xlabel(
        feature.replace('_', ' ').title()
    )

    plt.ylabel("Number of Patients")

    plt.xticks(rotation=45)

    plt.legend(
        title="Readmission",
        labels=["No", "Yes"]
    )

    plt.tight_layout()
    plt.show()
```

### 12. CORRELATION ANALYSIS


```python
# ============================================================
# 12. CORRELATION ANALYSIS
# ============================================================

print("=" * 60)
print("CORRELATION ANALYSIS")
print("=" * 60)

# Select numerical columns including target
correlation_data = df.select_dtypes(
    include=["int64", "float64"]
)

correlation_matrix = correlation_data.corr()

display(correlation_matrix.round(2))

# Correlation heatmap
plt.figure(figsize=(12, 8))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0
)

plt.title("Correlation Heatmap")

plt.tight_layout()
plt.show()
```

### 13. CORRELATION WITH TARGET


```python
# ============================================================
# 13. CORRELATION WITH TARGET
# ============================================================

print("=" * 60)
print("FEATURE CORRELATION WITH READMISSION")
print("=" * 60)

target_correlation = (
    correlation_matrix["label"]
    .sort_values(ascending=False)
)

display(target_correlation)
```

### 14. OUTLIER ANALYSIS


```python
# ============================================================
# 14. OUTLIER ANALYSIS
# ============================================================

print("=" * 60)
print("OUTLIER ANALYSIS")
print("=" * 60)

for feature in numerical_features:

    plt.figure(figsize=(8, 3))

    sns.boxplot(
        x=df[feature]
    )

    plt.title(
        f"Outlier Analysis: {feature.replace('_', ' ').title()}"
    )

    plt.xlabel(
        feature.replace('_', ' ').title()
    )

    plt.tight_layout()
    plt.show()
```

### 15. IQR-BASED OUTLIER COUNT


```python
# ============================================================
# 15. IQR-BASED OUTLIER COUNTS
# ============================================================

print("=" * 60)
print("IQR-BASED OUTLIER COUNTS")
print("=" * 60)

outlier_summary = []

for feature in numerical_features:

    Q1 = df[feature].quantile(0.25)
    Q3 = df[feature].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[
        (df[feature] < lower_bound) |
        (df[feature] > upper_bound)
    ]

    outlier_summary.append({
        "Feature": feature,
        "Q1": Q1,
        "Q3": Q3,
        "IQR": IQR,
        "Lower Bound": lower_bound,
        "Upper Bound": upper_bound,
        "Outlier Count": len(outliers)
    })

outlier_df = pd.DataFrame(outlier_summary)

display(outlier_df)
```

### 16. READMISSION RATE BY IMPORTANT NUMERICAL FEATURES


```python
# ============================================================
# 16. READMISSION RATE BY IMPORTANT NUMERICAL FEATURES
# ============================================================

print("=" * 60)
print("READMISSION RATE ANALYSIS")
print("=" * 60)

# Previous readmissions
if "prev_readmissions" in df.columns:

    readmission_by_previous = (
        df.groupby("prev_readmissions")["label"]
        .mean()
        .mul(100)
        .round(2)
    )

    print("\nReadmission Rate by Previous Readmissions:")
    display(readmission_by_previous)


# Comorbidities
if "comorbidities_count" in df.columns:

    readmission_by_comorbidity = (
        df.groupby("comorbidities_count")["label"]
        .mean()
        .mul(100)
        .round(2)
    )

    print("\nReadmission Rate by Comorbidities:")
    display(readmission_by_comorbidity)


# Length of stay
if "length_of_stay" in df.columns:

    readmission_by_stay = (
        df.groupby("length_of_stay")["label"]
        .mean()
        .mul(100)
        .round(2)
    )

    print("\nReadmission Rate by Length of Stay:")
    display(readmission_by_stay)
```

### 17. CHECK RARE CATEGORIES


```python
# ============================================================
# 17. CHECK RARE CATEGORIES
# ============================================================

print("=" * 60)
print("RARE CATEGORY CHECK")
print("=" * 60)

for feature in categorical_features:

    counts = df[feature].value_counts()

    print(f"\n{feature}")

    display(
        counts[counts < len(df) * 0.01]
    )
```

```python
# ============================================================
# 18. KEY EDA FINDINGS
# ============================================================

print("=" * 60)
print("KEY EDA FINDINGS")
print("=" * 60)

print("""
1. Dataset structure and feature types were examined.

2. Missing values and duplicate records were checked.

3. Numerical feature distributions were analyzed.

4. Categorical feature frequencies were analyzed.

5. Relationships between important features and readmission
   were explored.

6. Correlations between numerical features were examined.

7. Potential outliers were identified using the IQR method.

8. The target variable was checked for class imbalance.

9. These findings will guide feature selection and preprocessing.
""")
```

### 19. FINAL DATASET SUMMARY


```python
# ============================================================
# 19. FINAL DATASET SUMMARY
# ============================================================

print("=" * 60)
print("FINAL EDA SUMMARY")
print("=" * 60)

print("Dataset Shape:", df.shape)

print("Numerical Features:", len(numerical_features))
print("Categorical Features:", len(categorical_features))

print(
    "Missing Values:",
    df.isnull().sum().sum()
)

print(
    "Duplicate Records:",
    df.duplicated().sum()
)

print("\nTarget Distribution:")
display(
    df["label"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)
```
